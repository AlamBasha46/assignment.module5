 **1) List out the features of HTML5**

 **-->** HTML5 introduced a range of features to enhance web development, multimedia, and interactivity. Below is a concise list of its key features:

1. **Semantic Elements**: New tags like `<header>`, `<footer>`, `<article>`, `<section>`, `<nav>`, and `<aside>` for better structure and accessibility.
2. **Multimedia Support**:
   - `<audio>` and `<video>` elements for native audio and video playback without plugins.
   - Support for multiple formats (e.g., MP4, WebM, OGG).
3. **Canvas and SVG**:
   - `<canvas>` for dynamic 2D graphics and animations via JavaScript.
   - Scalable Vector Graphics (SVG) for resolution-independent graphics.
4. **Form Enhancements**:
   - New input types: `email`, `url`, `date`, `time`, `range`, `color`, etc.
   - Attributes like `placeholder`, `required`, `pattern`, and `autocomplete`.
   - Native form validation without JavaScript.
5. **Geolocation API**: Allows websites to access user location with permission.
6. **Web Storage**:
   - `localStorage` for persistent key-value storage.
   - `sessionStorage` for per-session data.
7. **WebSockets**: Enables real-time, bidirectional communication between client and server.
8. **Web Workers**: Runs JavaScript in background threads for improved performance.
9. **Drag and Drop API**: Native support for drag-and-drop functionality.
10. **Offline Web Applications**:
    - Application Cache (now deprecated) and Service Workers for offline access.
    - Manifest file for caching resources.
11. **CSS3 Integration**: Enhanced styling capabilities with transitions, animations, and flexbox.
12. **Cross-Origin Resource Sharing (CORS)**: Securely allows resource sharing across domains.
13. **ContentEditable**: Allows users to edit content directly within HTML elements.
14. **History API**: Manipulates browser history for single-page applications.
15. **Improved Accessibility**: Better support for ARIA (Accessible Rich Internet Applications) roles.
16. **Microdata**: Embeds machine-readable data in HTML using schema.org vocabularies.
17. **Responsive Images**:
    - `<picture>` element and `srcset` attribute for adaptive image loading.
18. **MathML**: Support for mathematical expressions in browsers.
19. **Custom Data Attributes**: `data-*` attributes for storing custom metadata.
20. **Backward Compatibility**: Graceful degradation for older browsers.


**2) What are HTML Entities? List out 5 commonly used HTML entities.**

**-->** **HTML Entities** are special codes used in HTML to represent characters that have specific meanings in HTML syntax (e.g., `<`, `>`) or are difficult to type directly (e.g., symbols like ©). They begin with an ampersand (`&`) and end with a semicolon (`;`), either using a name (e.g., `&amp;`) or a numerical code (e.g., `&#38;`). Entities ensure proper rendering of characters and prevent parsing issues in HTML.

### 5 Commonly Used HTML Entities:
1. `&amp;` or `&#38;` - Represents the ampersand (`&`).
2. `&lt;` or `&#60;` - Represents the less-than sign (`<`).
3. `&gt;` or `&#62;` - Represents the greater-than sign (`>`).
4. `&quot;` or `&#34;` - Represents the double quote (`"`).
5. `&copy;` or `&#169;` - Represents the copyright symbol (`©`).

**3) Define accessibility in the context of web development. Discuss why it's essential to create accessible websites and how it benefits different user group**

**-->**  **Definition of Accessibility**: Web accessibility in web development means creating websites and applications that are usable by all, including individuals with visual, auditory, motor, cognitive, or neurological disabilities. It follows **WCAG** principles to ensure content is perceivable, operable, understandable, and robust, using features like semantic HTML and assistive technology support.

**Why Essential**: Accessibility fosters inclusivity, ensuring equal access to digital resources. It’s required for legal compliance (e.g., ADA, EU Web Accessibility Directive) to avoid lawsuits. It also broadens market reach, as 15-20% of people have disabilities, and enhances SEO and user experience for all.

**Benefits for User Groups**:
- **Visual Impairments**: Alt text for images and screen reader compatibility allow blind or low-vision users to navigate content.
- **Hearing Impairments**: Captions and transcripts make audio/video accessible to deaf users.
- **Motor Impairments**: Keyboard navigation and voice controls support users with limited mobility.
- **Cognitive Disabilities**: Simple language and consistent layouts help users with autism or dyslexia.
- **Older Adults**: Large text and intuitive design aid aging users with declining abilities.
- **General Users**: Responsive design and clear navigation improve usability for everyone, including those with temporary limitations (e.g., broken arm).

HTML5 enhances accessibility with semantic elements (`<nav>`, `<article>`), ARIA attributes for dynamic content, native form validation, and multimedia elements (`<video>`, `<audio>`) with caption support. Prioritizing accessibility creates inclusive, compliant, and user-friendly websites.

**4) List any 3 ways which help us in improving the accessibility of HTML**

**-->** Three effective ways to improve the accessibility of HTML:

1. **Use Semantic HTML Elements**: Utilize HTML5 semantic tags like `<header>`, `<nav>`, `<main>`, `<article>`, and `<footer>` to provide a clear structure. These elements help screen readers and assistive technologies understand the page’s layout and content, improving navigation for users with visual or cognitive impairments.

2. **Provide Alternative Text for Images**: Add descriptive `alt` attributes to `<img>` tags (e.g., `<img src="dog.jpg" alt="A golden retriever playing in a park">`). This ensures that screen readers convey the image’s content or function to blind or low-vision users, enhancing their understanding of the page.

3. **Ensure Keyboard Navigability**: Design HTML to support full keyboard interaction by using focusable elements (e.g., `<a>`, `<button>`) and avoiding JavaScript that traps focus. Add ARIA attributes like `aria-label` or `aria-hidden` where needed to clarify functionality, enabling users with motor impairments to navigate without a mouse.
